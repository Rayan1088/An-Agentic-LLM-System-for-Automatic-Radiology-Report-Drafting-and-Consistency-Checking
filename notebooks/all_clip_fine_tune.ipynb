
CLIP Evaluation Metrics
1. Recall@k: [Is the correct report present in the first "K retrieved 
	reports?] = [queries whose GT is in Top-k / total queries] in
	our case total test images = 600 =  total queries. Example: 
	we have Top-5 report [re-5, re-18, re-3, re-20, re-30] so for 	that particular Image Ground Truth Report is re-3, for that
	Recall@1=0, Recall@2=0, Recall@3=1, Recall@4=1, Recall@5=1. 
	Means in top 3, 4 and 5 the ground truth report is present.

2. Mean Reciprocal Rank (MRR):
	Example: Let we have total 3 test images,
	for Image-1, Correct report at rank 1. For that MMR=1/1=1
	for Image-2, Correct report at rank 4. For that MMR=1/4=0.25
	for Image-3, Correct report at rank 2. For that MMR=1/2=0.50
	So Average Reciprocal Rank=(1+0.25+0.50)/3
	[N.B. Higher value is Better]

3. Mean Rank: Tell us where the correct report is ranked on average 	across all query images. Example: We have total 3 test images 	for Image-1, correct report rank in Top-5 is 2
	for Image-2, correct report rank in Top-5 is 4
	for Image-3, correct report rank in Top-5 is 3
	So Mean Rank = (2+4+4)/3
	[N.B. For retrieval system "Lover Mean Rank Value Is Better".]

4. Median Rank: What is middle rank of the correct reportr across all 
	query images? When we have Big "Mean Rank" value then we 	will 	consider "Median Rank" for that experiment. 
	[N.B. For retrieval system "Lover Median Rank Value Is Better".]
